### Dynamic Capabilities

```python
dbutils.widgets.text("file_name","")
```

```python
file_name = dbutils.widgets.get("file_name")
```

```python
print(file_name)
```

### Data Reading

```python
df = spark.readStream.format("cloudFiles")\
        .option("cloudFiles.format","parquet")\
        .option("cloudFiles.schemaLocation",f"abfss://bronze@storageaccount03five.dfs.core.windows.net/checkpoint_{file_name}")\
        .load(f"abfss://sourcecontainer035@storageaccount03five.dfs.core.windows.net/{file_name}") 
```

### Data Writing

```python
query = (df.writeStream.format("parquet")\
     .option("checkpointLocation",f"abfss://bronze@storageaccount03five.dfs.core.windows.net/checkpoint_{file_name}")\
     .option("path",f"abfss://bronze@storageaccount03five.dfs.core.windows.net/{file_name}")\
     .trigger(once=True)\
     .start()) 
     
query.awaitTermination()
```
